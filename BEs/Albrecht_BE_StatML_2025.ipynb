```python
# import functions used for linear regression
from sklearn.linear_model import LinearRegression, LassoCV
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import KFold, LeaveOneOut
import statsmodels.api as sm
```

```python
"""
function to fit data for each brand.
Args :
    lr      : LinearRegression object
    x       : array, input data
    y       : array, output data
    plot    : bool, wether to plot the results
    info    : a string to add on plots
Returns :
    y_hat   : predicted y
    msq     : mean square error
    R2      : the R² score
lr object having been previously instanciated, coefficient are stored inside.
"""
def LinReg(lr, x, y, plot=True, xleg_yleg=["X data", "Y data"], info:str=""):
    # copy and expand dimensions
    x0 = np.zeros(x.shape).reshape(-1, 1)
    x0[:, 0] = x[:]
    y0 = np.zeros(y.shape)
    y0[:] = y[:]
    # fit model
    lr.fit(x0, y0)
    # compute predicted y using model
    y_hat = lr.predict(x0)
    # compute mean square error and Pearson coeff squared (R²)
    msq = mean_squared_error(y0, y_hat)
    R2 = r2_score(y0, y_hat)
    if plot :
        # plot
        plt.figure(dpi=150)
        plt.scatter(x0, y0, color="orchid", marker="o", label=f"data {info}")
        plt.plot(x0, y_hat, color="dodgerblue", marker="", label=f"linear reg {info}\nintercept: {SmartRounder(lr.intercept_)}\nslope: {SmartRounder(lr.coef_[0])}")
        plt.xlabel(xleg_yleg[0])
        plt.ylabel(xleg_yleg[1])
        plt.grid()
        plt.legend()
        plt.title(f"Computed linear regression for {info}, R²={round(R2, 2)}")
        plt.show()
    return y_hat, msq, R2

"""
Function that simply displays the result of a linear regression.
Ars :
    msq : mean squared error
    lr : LinearRegression object
"""
def DisplayLinearResults(lr, msq, R, x="X", y="Y"):
    print("\n **** \n")
    print(f"{x}  wrt  {y}")
    print(f"Mean square error for first linear regression is {msq}.")
    print(f"Mean square rooted error for first linear regression is {np.sqrt(msq)}.")
    print(f"Slope of model (eg impact of {x} on {y}) is roughly {np.round(lr.coef_[0]*100, 1)} %.")
    print(f"Corresponding R² is {R}")
    return None


"""
return x with precision+1 digits. For display on graphs.
"""
def SmartRounder(x, precision=3):
    try :
        # check for iterable of length 0
        if hasattr(x, '__iter__') :
            if len(x) == 1 :
                x = x[0]
        power = int(np.floor(np.log(x)/np.log(10)))
        return np.round(x / 10**power, precision)*10**power
    except :
        print(f"could not round {x}")
        return x
```

```python

# instanciate a LinearRegression object for each brand
lr1 = LinearRegression()
lr2 = LinearRegression()

# solve and plot
_, msq1, R2_1 = LinReg(lr1, c_brand_1, e_brand_1, xleg_yleg=["Concentration", "Efficiency"], plot=True, info="brand_1")
_, msq2, R2_2 = LinReg(lr2, c_brand_2, e_brand_2, xleg_yleg=["Concentration", "Efficiency"], plot=True, info="brand_2")
# print results
DisplayLinearResults(lr1, msq1, R2_1, x="Concentration", y="Efficiency")
DisplayLinearResults(lr2, msq2, R2_2, x="Concentration", y="Efficiency")
```

```python
# Les deux indicateurs que l'on peut retenir sont la p-value et le R². 

"""
Function to determine wether x and y have a statistically significant linear relation.
Args :
    x : a 1D array of input data
    y : a 1D array of output data
Returns :
    relation :  (boolean) wether x and y are in a linear relation, based on p_value
"""
def StatisticalRelation(x, y, toprint=True):
    # linear regression
    lr = LinearRegression()
    y_hat, msq, r2  = LinReg(lr, x, y, plot=False)
    # estimate p_value
    Xsm = sm.add_constant(x.reshape(-1, 1))
    fitted_est = sm.OLS(y, Xsm).fit()
    p_value = fitted_est.pvalues[1] # p_value = p_intercept, p_slope
    # print
    if toprint : print(f"p_value is {p_value} and R² is {r2}.")
    return p_value < 0.05

print("\n***\nsensor 01")
print(StatisticalRelation(X[:, 0], y))
print("\n***\nsensor 12")
print(StatisticalRelation(X[:, 11], y))
print("\n***\nsensor 15")
print(StatisticalRelation(X[:, 14], y))

```

Avant d'effectuer une régression linéaire régularisée, il est vivement conseillé d'effectuer une étape de centrage réduction des données. Pourquoi ?


```python
"""
Le centrage-réduction des données permet d'éviter qu'une variable d'entrée atteignant des valeurs beaucoup plus importantes ne supplante toutes les autres.
Cela permet également de mieux comprendre l'effet du coefficient β0 sur yi, car yi vaudra β0 quand xi vaut moy(xi).
"""
pass
```

```python
# define functions for outlier detection

def hij(x, i, j):
    n = len(x)
    m = np.mean(x)
    h = 1/n + (  (x[i] - m) * (x[j] - m)  ) / (np.sum((x - m)**2))
    return h

"""
Function that detects outliers by computing their leverage on prediction, and returns the indices of the ouliers.
Args :
    x : array of values in which outliers might be present
    y : array of out values
    p : detection parameter : number above p times the standard deviation will be outliered
    plot :      wether to plot
    xleg_yleg : x and y legends to display on the plot
Returns :
    indices : array of indices of the detected outliers. Empty array if no outliers.
    mask : the mask of the values to keep in x (those that are not outliers)
"""
def DetectOutliers(x, y, p=2, plot=True, xleg_yleg=["x", "y"]):
    ny = len(y)
    # initialize matrix of hij coeff
    H = np.zeros((ny, ny))
    # fill H with hij values
    # TODO : vectorize
    for i in range(ny):
        for j in range(ny):
            H[i, j] = hij(x, i, j)
    # signs are not needed
    H = np.abs(H)
    # compute mean and standard deviation of H
    Hmean = np.mean(H)
    Hstd = np.std(H.flatten())
    print(f"mean is {Hmean}, σ is {Hstd}")
    # outliers are xi whose corresponding hii have a value higher than usual (eg, p*Hstd)
    diag = np.diag(H)
    indices, = np.where(diag > Hmean + p*Hstd)
    mask = diag <= (Hmean + p*Hstd)

    if plot :
        # plot data
        plt.figure(dpi=90)
        plt.scatter(x, y, color="lightgreen", marker="o", label=f"Data")
        plt.scatter(x[indices], y[indices], color="tomato", marker="x", label=f"Outlier")
        plt.xlabel(xleg_yleg[0])
        plt.ylabel(xleg_yleg[1])
        plt.grid()
        plt.legend()
        plt.title(f"detected outliers for {xleg_yleg[0]} wrt to {xleg_yleg[1]}")
        plt.show()
        # plot hii repartition
        plt.figure(dpi=90)
        plt.hist(diag, bins=10)
        plt.xlabel("leverage coefficient")
        plt.ylabel("# data")
        plt.grid()
        plt.title("repartition of leverage coefficient")
        plt.show()
    return indices, mask
```

```python
# function for centering, reducing

"""
Function to reduce and center a 2D array.
Args :
    X :     The 2-D array whose columns are to be centered
Returns :
    X0 :    Centered reduced X (mean=0, variance=1 on each **column**)
"""
def CenterReduce(X): # multi-axis version
    X0 = np.zeros(X.shape)
    X1 = X.T # to perform computation on columns
    print(f"Reducing variable of shape {X.shape}")
    m = np.mean(X1, axis=1) * np.ones(X.shape)
    s = np.var(X1, axis=1) * np.ones(X.shape)
    X0[:, :] = (X[:, :] - m) / s**2
    return X0
```

```python

# centering and reducing data
Xc = CenterReduce(Xm)

# select only statistically relevant data BEFORE lasso selection
only_relevant = True
if only_relevant :
    # selection
    mask_var = []
    for j in range(X.shape[1]):
        mask_var.append(StatisticalRelation(X[:, j], y, toprint=False))
    # transform X
    Xc = Xc[:, mask_var]

print(f"selected variables {np.arange(0, len(mask_var), 1)[mask_var]}")

```

```python

# lasso select

# choose alphas
alphas = np.logspace(-4, 1, 100)
#print(f" Possible alphas : {alphas}")
# alpha dict to store results
alphadict = {}
# choose k
K = 8
# kfold train
kf = KFold(n_splits = K , shuffle=False, random_state=None)
# compute splits on the dataset
kf.get_n_splits(Xc)

for alpha in alphas :
    mse_splits = []
    for i, (train_index, test_index) in enumerate(kf.split(Xc)):
        #print(f"Fold {i} for alpha {alpha}")
        #print(f"  Train: index={train_index}")
        #print(f"  Test:  index={test_index}")
        # effectively split data based on computed fold
        Xc_train = Xc[train_index]
        y_train = ym[train_index]
        Xc_test = Xc[test_index]
        y_test = ym[test_index]
        # train on split DEPENDING ON ALPHA
        lasso_regressor = Lasso(alpha=alpha)
        lasso_regressor.fit(Xc_train, y_train)
        y_pred_lasso = lasso_regressor.predict(Xc_test)
        # compute and store error
        mse_split = mean_squared_error(y_test, y_pred_lasso)
        mse_splits.append(mse_split)

    # mean error on splits
    mse_allsplits = np.mean(mse_splits)
    # store result
    alphadict.update({alpha : mse_allsplits})

alphasR, msqR = list(alphadict.keys()), list(alphadict.values())

plt.figure(dpi=130)
plt.plot(alphasR, msqR, color="dodgerblue", label="mean squared error")
plt.plot(alphasR, msqR, color="gold", linestyle='', marker="o", markevery=[np.argmin(msqR)], 
label=f"min : {round(np.min(msqR), 3)} for alpha={round(alphasR[np.argmin(msqR)], 3)}")
plt.legend()
plt.grid()
plt.show()
```
